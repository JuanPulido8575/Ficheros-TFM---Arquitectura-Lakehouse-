## Anonimización de datos 
(El texto Markdown siguiente debe llevarse a la sección de Marco Normativo del documento del TFM)

La anonimización de datos es crucial para reducir el riesgo inherente de identificación al que están sometidos los interesados del tratamientos de datos. Puesto que el caso de uso de este trabajo de fin de máster involucra necesariamente variables de carácter personal, en virtud del RGPD y la ley orgánica 3/2018 (NLOPD) que velan por la protección y seguridad de datos personales de terceros que sirvan para identificar indirecta o directamente y de forma unívoca una persona física, ha de elaborarse previamente un proceso sistemático que reduzca en la mayor medida posible el riesgo de reidentificación de los votantes que atente en contra de sus derechos y libertades, al tiempo que permita conservar la máxima utilidad posible de los datos, de modo que sigan siendo lo bastante fieles a los datos originales para las tareas de BI y BA que se han puesto como parte del objetivo del trabajo de fin de máster. De esta forma, dejan de ser personales y en consecuencia ya no estarían sujetos a la normativa de protección de datos (citar aquí: https://www.edps.europa.eu/system/files/2021-04/21-04-27_aepd-edps_anonymisation_en_5.pdf). Considerando los clústeres de campos para los datos de votantes y llamadas que se han obtenido en la sección de metodología junto a los campos potenciales que en un principio se consideraron para el caso de uso de uso de este trabajo, se ha mapeado a cada clúster, una técnica específica de anonimización. A saber:

Esto es fundamental para relacionar consistentemente la tabla de votantes con la tabla de llamadas.

#### Anonimización en Voters

```python
import hashlib

def anonymize_value(value: str, salt: str = SALT, length: int = None) -> str:
    """
    Anonimiza un valor con SHA-256 + salt y devuelve un hash hexadecimal.
    Si length está especificado, devuelve solo los primeros `length` caracteres.
    """
    digest = hashlib.sha256((salt + str(value)).encode("utf-8")).hexdigest()
    return digest if length is None else digest[:length]

def anonymize_column(df, column, new_column=None, salt=SALT, length=None):
    if new_column is None:
        new_column = f"{column}_anon"
    df[new_column] = (
        df[column]
        .astype(str)
        .apply(lambda x: anonymize_value(x, salt, length=length))
    )
    return df

# Configurable salt (keep it secret and constant)
SALT = "3f8b2c9a-7d4e-4f1a-b2c0-1234567890ab"

# En este punto el reto consiste en recorrer todas las carpetas de votantes y llamadas para pseudo-anonimizar fichero por fichero un conjunto de
# campos donde cada conjunto representa un mismo atributo de la realidad. Por ejemplo



# Ejemplo: sólo 16 caracteres de hash para el teléfono
df = anonymize_column(df, "phone", new_column="phone_anon16")
```
